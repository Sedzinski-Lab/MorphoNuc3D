```python
from tifffile import imread, imwrite
from cellpose import models, train
import numpy as np
import napari
from czifile import imread as czi_imread
from n2v.models import N2V
```

# check model


```python
new_model_path = '../models/models/cellpose_n2v_id468'
```

```python
image_path = '../DATA/260330_ID479_laminB3/ST28/ST28_emb1_2_2026_03_30__09_41_38.czi'
image_raw = czi_imread(image_path)
img_nuclei_raw = image_raw[0, 0, 2, 0, :, :, :,0]  # (Z, Y, X)
```

```python
img_nuclei = imread('../DATA/260330_ID479_laminB3/ST28/ST28_emb1_2_2026_03_30__09_41_38_n2v_3d.tif')  # (Z, Y, X)
```

```python
fine_model = models.CellposeModel(gpu=True, pretrained_model=str(new_model_path))

mask, flows, styles = fine_model.eval(
    img_nuclei,
    do_3D=False,
    z_axis=0,
    anisotropy=0.3 / 0.1,
    normalize=True,
    diameter=50,
    flow_threshold=0.2,       
    cellprob_threshold=0.4,   # this is the main threshold to tune in 3D
    stitch_threshold=0.3,
    min_size=30,
)
print(f"nuclei found: {mask.max()}")
```

```python
viewer = napari.Viewer()
viewer.add_image(img_nuclei, name='nuclei')
viewer.add_image(img_nuclei_raw, name='nuclei_raw')
viewer.add_labels(mask, name='predicted masks denoised')
```
